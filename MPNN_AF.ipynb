```python
#@title Run ProteinMPNN to design new sequences for PDBs in Drive folder

import warnings, os, re, glob
import pandas as pd
from google.colab import drive

warnings.simplefilter(action='ignore', category=FutureWarning)

drive.mount('/content/drive')

# USER OPTIONS

#@markdown #### ProteinMPNN options
model_name = "v_48_020" #@param ["v_48_002", "v_48_010", "v_48_020", "v_48_030"]

#@markdown #### Input Options
pdb_folder = "/content/drive/MyDrive/RFdiffusion_outputs" #@param {type:"string"}
chains = "A,B" #@param {type:"string"}
homooligomer = False #@param {type:"boolean"}

#@markdown #### Replace RFdiffusion poly-G target before ProteinMPNN
replace_polyG_target = True #@param {type:"boolean"}
target_chain_for_mpnn = "B" #@param {type:"string"}
target_real_sequence = "" #@param {type:"string"}
target_real_sequence_file = "/content/drive/MyDrive/AF-Q13148-F1-model_v6.pdb" #@param {type:"string"}
target_real_sequence_file_chain = "A" #@param {type:"string"}
target_real_sequence_range = "310-350" #@param {type:"string"}
#@markdown - for PDB input, `target_real_sequence_range` uses PDB residue numbers, example: `310-350`

#@markdown #### Design constraints
fix_pos = "B" #@param {type:"string"}
#@markdown - specify which positions to keep fixed in the sequence (example: `1,2-10`)
#@markdown - chain-specific examples: `A1-10,B1-20`; entire chain example: `A`
inverse = False #@param {type:"boolean"}
#@markdown - inverse the `fix_pos` selection: positions listed are designed, all others fixed
rm_aa = "" #@param {type:"string"}
#@markdown - specify amino acid(s) to exclude (example: `C,A,T`)

#@markdown #### Design Options
num_seqs = 32 #@param ["32", "64", "128", "256", "512", "1024"] {type:"raw"}
sampling_temp = 0.1 #@param ["0.0001", "0.1", "0.15", "0.2", "0.25", "0.3", "0.5", "1.0"] {type:"raw"}
resume_mpnn = True #@param {type:"boolean"}
force_rerun_mpnn = False #@param {type:"boolean"}

# Save all results in the current Drive PDB folder.
output_dir = os.path.join(pdb_folder, "proteinmpnn_alphafold_results")
os.makedirs(output_dir, exist_ok=True)

# Clean user options.
chains = re.sub("[^A-Za-z]+", ",", chains).strip(",")

if fix_pos == "":
    fix_pos = None

rm_aa = ",".join(list(re.sub("[^A-Z]+", "", rm_aa.upper())))
if rm_aa == "":
    rm_aa = None

AA3_TO_1 = {
    "ALA": "A", "ARG": "R", "ASN": "N", "ASP": "D", "CYS": "C",
    "GLN": "Q", "GLU": "E", "GLY": "G", "HIS": "H", "ILE": "I",
    "LEU": "L", "LYS": "K", "MET": "M", "PHE": "F", "PRO": "P",
    "SER": "S", "THR": "T", "TRP": "W", "TYR": "Y", "VAL": "V",
    "SEC": "U", "PYL": "O",
}
AA1_TO_3 = {
    "A": "ALA", "R": "ARG", "N": "ASN", "D": "ASP", "C": "CYS",
    "Q": "GLN", "E": "GLU", "G": "GLY", "H": "HIS", "I": "ILE",
    "L": "LEU", "K": "LYS", "M": "MET", "F": "PHE", "P": "PRO",
    "S": "SER", "T": "THR", "W": "TRP", "Y": "TYR", "V": "VAL",
    "U": "SEC", "O": "PYL",
}

def clean_seq(seq):
    return str(seq).replace("/", "").replace(":", "").replace(" ", "").replace("\n", "").upper()

def parse_residue_range(residue_range):
    residue_range = str(residue_range).strip()
    if residue_range == "":
        return None, None
    if "-" not in residue_range:
        resi = int(residue_range)
        return resi, resi
    start, end = residue_range.split("-", 1)
    return int(start), int(end)

def resolve_existing_file(path):
    path = str(path).strip()
    if os.path.isfile(path):
        return path

    basename = os.path.basename(path)
    search_dirs = []
    dirname = os.path.dirname(path)
    if dirname:
        search_dirs.append(dirname)
    search_dirs.append(pdb_folder)

    seen_dirs = set()
    for folder in search_dirs:
        if not folder or folder in seen_dirs or not os.path.isdir(folder):
            continue
        seen_dirs.add(folder)
        for candidate in glob.glob(os.path.join(folder, "*")):
            if os.path.basename(candidate).lower() == basename.lower() and os.path.isfile(candidate):
                print("Resolved target sequence file:", candidate)
                return candidate

    raise FileNotFoundError(f"target_real_sequence_file not found: {path}")

def sequence_from_fasta(path):
    seq = []
    with open(path) as handle:
        for line in handle:
            line = line.strip()
            if line and not line.startswith(">"):
                seq.append(line)
    return clean_seq("".join(seq))

def apply_sequence_range(seq, residue_range):
    residue_range = str(residue_range).strip()
    if residue_range == "":
        return seq
    if "-" not in residue_range:
        idx = int(residue_range) - 1
        return seq[idx:idx + 1]
    start, end = residue_range.split("-", 1)
    start = int(start) - 1
    end = int(end)
    return seq[start:end]

def sequence_from_pdb_chain(path, chain_id, residue_range=""):
    start_resi, end_resi = parse_residue_range(residue_range)
    residues = []
    seen = set()
    with open(path) as handle:
        for line in handle:
            if not line.startswith(("ATOM", "HETATM")) or len(line) < 27:
                continue
            chain = line[21].strip()
            if chain != chain_id:
                continue
            try:
                resi_num = int(line[22:26])
            except ValueError:
                continue
            if start_resi is not None and not (start_resi <= resi_num <= end_resi):
                continue
            resid = (chain, resi_num, line[26].strip())
            if resid in seen:
                continue
            seen.add(resid)
            residues.append(AA3_TO_1.get(line[17:20].strip(), "X"))
    return clean_seq("".join(residues))

def get_target_real_sequence_for_mpnn():
    if str(target_real_sequence).strip():
        seq = clean_seq(target_real_sequence)
        print("Using target_real_sequence from text input.")
        return seq

    path = resolve_existing_file(target_real_sequence_file)
    lower = path.lower()
    if lower.endswith((".fa", ".fasta", ".faa", ".txt")):
        seq = apply_sequence_range(sequence_from_fasta(path), target_real_sequence_range)
    elif lower.endswith((".pdb", ".ent")):
        seq = sequence_from_pdb_chain(path, target_real_sequence_file_chain, target_real_sequence_range)
    else:
        raise ValueError("target_real_sequence_file must be FASTA, TXT, or PDB.")

    print("Using MPNN target sequence from file:", path)
    print("Target file chain:", target_real_sequence_file_chain)
    print("Target sequence range:", target_real_sequence_range if str(target_real_sequence_range).strip() else "full")
    print("Target sequence for fixed chain:", seq)
    return seq

def split_seq_by_lengths(seq, lengths):
    seq = clean_seq(seq)
    pieces = []
    start = 0
    for length in lengths:
        pieces.append(seq[start:start + length])
        start += length
    return pieces

def format_seq_by_lengths(seq, lengths):
    return "/".join(split_seq_by_lengths(seq, lengths))

def pdb_chain_sequence(path, chain_id):
    return sequence_from_pdb_chain(path, chain_id, "")

def replace_pdb_chain_sequence(input_pdb, output_pdb, chain_id, new_sequence):
    new_sequence = clean_seq(new_sequence)
    residue_order = []
    seen = set()
    with open(input_pdb) as handle:
        lines = handle.readlines()

    for line in lines:
        if not line.startswith(("ATOM", "HETATM")) or len(line) < 27:
            continue
        if line[21].strip() != chain_id:
            continue
        resid = (line[21].strip(), line[22:26].strip(), line[26].strip())
        if resid not in seen:
            seen.add(resid)
            residue_order.append(resid)

    if len(residue_order) != len(new_sequence):
        raise ValueError(
            f"Cannot replace chain {chain_id} in {input_pdb}: "
            f"PDB chain has {len(residue_order)} residues but target sequence has {len(new_sequence)} residues."
        )

    resid_to_resname = {
        resid: AA1_TO_3.get(aa, "UNK")
        for resid, aa in zip(residue_order, new_sequence)
    }

    os.makedirs(os.path.dirname(output_pdb), exist_ok=True)
    with open(output_pdb, "w") as out:
        for line in lines:
            if line.startswith(("ATOM", "HETATM")) and len(line) >= 27 and line[21].strip() == chain_id:
                resid = (line[21].strip(), line[22:26].strip(), line[26].strip())
                resname = resid_to_resname[resid]
                line = line[:17] + f"{resname:>3}" + line[20:]
            out.write(line)
    return output_pdb

target_sequence_for_mpnn = get_target_real_sequence_for_mpnn() if replace_polyG_target else ""
if replace_polyG_target:
    print("MPNN fixed target chain:", target_chain_for_mpnn)
    print("MPNN fixed target length:", len(target_sequence_for_mpnn))

pdb_files = sorted(glob.glob(os.path.join(pdb_folder, "*.pdb")))

if len(pdb_files) == 0:
    raise FileNotFoundError(f"No .pdb files found in: {pdb_folder}")

if "mpnn_model" not in dir() or "model_name_" not in dir() or model_name_ != model_name:
    mpnn_model = mk_mpnn_model(model_name)
    model_name_ = model_name

mpnn_runs = []
all_results = []

for pdb_path in pdb_files:
    pdb_name = os.path.splitext(os.path.basename(pdb_path))[0]
    print(f"\nDesigning sequences for: {pdb_name}")

    fasta_path = os.path.join(output_dir, f"{pdb_name}_design.fasta")
    csv_path = os.path.join(output_dir, f"{pdb_name}_mpnn_results.csv")

    if resume_mpnn and not force_rerun_mpnn and os.path.isfile(csv_path):
        cached_df = pd.read_csv(csv_path)
        required_cols = {"score", "seqid", "seq", "chains", "fix_pos", "inverse", "rm_aa", "model_name", "target_chain_for_mpnn", "target_sequence_for_mpnn"}
        cache_matches_options = False
        if len(cached_df) >= num_seqs and required_cols.issubset(cached_df.columns):
            cache_matches_options = (
                str(cached_df["chains"].iloc[0]) == str(chains)
                and str(cached_df["fix_pos"].iloc[0]) == str(fix_pos)
                and str(cached_df["inverse"].iloc[0]).lower() == str(bool(inverse)).lower()
                and str(cached_df["rm_aa"].iloc[0]) == str(rm_aa)
                and str(cached_df["model_name"].iloc[0]) == str(model_name)
                and str(cached_df["target_chain_for_mpnn"].iloc[0]) == str(target_chain_for_mpnn)
                and str(cached_df["target_sequence_for_mpnn"].iloc[0]) == str(target_sequence_for_mpnn)
            )
        if cache_matches_options:
            print(f"Skipping ProteinMPNN for {pdb_name}; using cached CSV with matching options: {csv_path}")
            cached_df = cached_df.head(num_seqs).copy()
            out = {
                "score": cached_df["score"].to_numpy(),
                "seqid": cached_df["seqid"].to_numpy(),
                "seq": cached_df["seq"].astype(str).tolist(),
            }
            mpnn_runs.append({
                "pdb_name": pdb_name,
                "pdb_path": cached_df["pdb_path_for_mpnn"].iloc[0] if "pdb_path_for_mpnn" in cached_df.columns else pdb_path,
                "out": out,
                "from_cache": True,
            })
            all_results.append(cached_df)
            continue
        print(f"Cached MPNN CSV missing option metadata or options changed; rerunning: {csv_path}")

    pdb_path_for_mpnn = pdb_path
    if replace_polyG_target:
        corrected_pdb_dir = os.path.join(output_dir, "input_pdb_with_real_target")
        pdb_path_for_mpnn = os.path.join(corrected_pdb_dir, f"{pdb_name}_target{target_chain_for_mpnn}_real_sequence.pdb")
        replace_pdb_chain_sequence(
            input_pdb=pdb_path,
            output_pdb=pdb_path_for_mpnn,
            chain_id=target_chain_for_mpnn,
            new_sequence=target_sequence_for_mpnn,
        )
        print("Original target chain sequence:", pdb_chain_sequence(pdb_path, target_chain_for_mpnn))
        print("Corrected target chain sequence:", pdb_chain_sequence(pdb_path_for_mpnn, target_chain_for_mpnn))
        print("ProteinMPNN input PDB:", pdb_path_for_mpnn)

    mpnn_model.prep_inputs(
        pdb_filename=pdb_path_for_mpnn,
        chain=chains,
        homooligomer=homooligomer,
        fix_pos=fix_pos,
        inverse=inverse,
        rm_aa=rm_aa,
        verbose=True
    )

    out = mpnn_model.sample(
        num=num_seqs // 32,
        batch=32,
        temperature=sampling_temp,
        rescore=homooligomer
    )

    selected_chains_for_mpnn = [c for c in chains.split(",") if c]
    chain_lengths_for_mpnn = [len(pdb_chain_sequence(pdb_path_for_mpnn, c)) for c in selected_chains_for_mpnn]
    print("ProteinMPNN chain order:", "/".join(selected_chains_for_mpnn))
    print("ProteinMPNN chain lengths:", chain_lengths_for_mpnn)

    mpnn_runs.append({
        "pdb_name": pdb_name,
        "pdb_path": pdb_path_for_mpnn,
        "original_pdb_path": pdb_path,
        "out": out
    })

    rows = []

    with open(fasta_path, "w") as fasta:
        for n in range(num_seqs):
            header = f'>{pdb_name}_n{n}_score:{out["score"][n]:.3f}_seqid:{out["seqid"][n]:.3f}'
            seq = out["seq"][n]

            fasta.write(header + "\n" + seq + "\n")

            seq_by_chain = format_seq_by_lengths(seq, chain_lengths_for_mpnn)

            print(header)
            print(seq)
            print("Sequence by chain:")
            print(seq_by_chain)

            rows.append([
                pdb_name, n, out["score"][n], out["seqid"][n], seq, seq_by_chain,
                chains, fix_pos, inverse, rm_aa, model_name,
                target_chain_for_mpnn, target_sequence_for_mpnn, pdb_path_for_mpnn,
            ])

    df = pd.DataFrame(
        rows,
        columns=[
            "pdb", "design_index", "score", "seqid", "seq", "seq_by_chain", "chains", "fix_pos", "inverse", "rm_aa", "model_name",
            "target_chain_for_mpnn", "target_sequence_for_mpnn", "pdb_path_for_mpnn",
        ]
    )
    df.to_csv(csv_path, index=False)
    all_results.append(df)

all_df = pd.concat(all_results, ignore_index=True)
all_df.to_csv(os.path.join(output_dir, "all_mpnn_results.csv"), index=False)

data_table.DataTable(all_df.round(3))
```

```python
#@title AF2-IG Complex -> AF2 Monomer -> Rosetta Interface Quality

#@markdown Run this after the ProteinMPNN cell. This cell defines the full triage function.
run_pipeline_now = True #@param {type:"boolean"}
max_designs_per_pdb = 4 #@param {type:"integer"}
designed_chain = "A" #@param {type:"string"}
target_chains = "B" #@param {type:"string"}
target_chain = "B" #@param {type:"string"}
target_real_sequence = "" #@param {type:"string"}
target_real_sequence_file = "/content/drive/MyDrive/AF-Q13148-F1-model_v6.pdb" #@param {type:"string"}
target_real_sequence_file_chain = "A" #@param {type:"string"}
target_real_sequence_range = "310-350" #@param {type:"string"}
#@markdown - use `target_real_sequence` directly, or provide a FASTA/PDB path in `target_real_sequence_file`
#@markdown - for PDB input, `target_real_sequence_range` uses PDB residue numbers, example: `310-350`
#@markdown - leave blank to score the last selected chain as the designed chain
complex_num_models = 1 #@param ["1","2","3","4","5"] {type:"raw"}
complex_num_recycles = 1 #@param ["0","1","2","3"] {type:"raw"}
monomer_num_models = 1 #@param ["1","2","3","4","5"] {type:"raw"}
monomer_num_recycles = 1 #@param ["0","1","2","3"] {type:"raw"}
run_monomer_af2 = True #@param {type:"boolean"}
run_interface_quality = True #@param {type:"boolean"}
resume_pipeline = True #@param {type:"boolean"}
force_rerun_pipeline = False #@param {type:"boolean"}

import os, glob, math, shutil, subprocess, inspect, gc
import numpy as np
import pandas as pd
import jax
import jax.numpy as jnp
from colabdesign.af import mk_af_model

# Same JAX compatibility shim used in the AF cell.
if "a_min" not in inspect.signature(jnp.clip).parameters:
    _jnp_clip_pipeline = jnp.clip

    def _clip_compat_pipeline(a, a_min=None, a_max=None, out=None, min=None, max=None):
        if min is None:
            min = a_min
        if max is None:
            max = a_max
        return _jnp_clip_pipeline(a, min=min, max=max)

    jnp.clip = _clip_compat_pipeline

if "mpnn_runs" not in dir() or len(mpnn_runs) == 0:
    raise RuntimeError("No ProteinMPNN results found. Run the ProteinMPNN cell first.")

if "output_dir" not in dir():
    output_dir = os.path.join(
        "/content/drive/MyDrive/sequence_input_diffusion_example_outputs",
        "proteinmpnn_alphafold_results"
    )
    os.makedirs(output_dir, exist_ok=True)

af_data_dir = globals().get("af_data_dir", "/content")
params_dir = os.path.join(af_data_dir, "params")
params_tar = os.path.join(af_data_dir, "alphafold_params_2022-12-06.tar")
params_url = "https://storage.googleapis.com/alphafold/alphafold_params_2022-12-06.tar"

def required_af_model_names(use_multimer, use_templates=False):
    if use_multimer:
        return [f"model_{k}_multimer_v3" for k in [1, 2, 3, 4, 5]]
    if use_templates:
        return [f"model_{k}_ptm" for k in [1, 2]]
    return [f"model_{k}_ptm" for k in [1, 2, 3, 4, 5]]

def af_param_path(model_name):
    return os.path.join(params_dir, f"params_{model_name}.npz")

def ensure_af_params(use_multimer, use_templates=False):
    required = required_af_model_names(use_multimer, use_templates)
    missing = [name for name in required if not os.path.isfile(af_param_path(name))]
    print(f"AlphaFold params directory: {params_dir}")
    for name in required:
        status = "FOUND" if os.path.isfile(af_param_path(name)) else "MISSING"
        print(f"{status}: {af_param_path(name)}")

    if missing:
        print("Downloading missing AlphaFold params:", ", ".join(missing))
        os.makedirs(params_dir, exist_ok=True)
        if os.path.isfile(params_tar):
            os.remove(params_tar)
        subprocess.check_call("apt-get install aria2 -qq", shell=True)
        subprocess.check_call(
            f"aria2c -q -x 16 -d {af_data_dir} -o {os.path.basename(params_tar)} {params_url}",
            shell=True
        )
        subprocess.check_call(f"tar --overwrite -xf {params_tar} -C {params_dir}", shell=True)

    missing = [name for name in required if not os.path.isfile(af_param_path(name))]
    if missing:
        raise RuntimeError("Still missing AlphaFold params: " + ", ".join(missing))

    print("Using AlphaFold data_dir:", af_data_dir)
    return required

def pdb_chain_ids(pdb_path):
    seen = []
    with open(pdb_path) as handle:
        for line in handle:
            if line.startswith(("ATOM", "HETATM")) and len(line) > 21:
                chain = line[21].strip()
                if chain and chain not in seen:
                    seen.append(chain)
    return seen

def write_chain_pdb(input_pdb, chain_id, output_pdb):
    with open(input_pdb) as inp, open(output_pdb, "w") as out:
        for line in inp:
            if line.startswith(("ATOM", "HETATM")) and len(line) > 21 and line[21].strip() == chain_id:
                out.write(line)
        out.write("END\n")
    return output_pdb

def clean_seq(seq):
    return str(seq).replace("/", "").replace(":", "").replace(" ", "").replace("\n", "").upper()

AA3_TO_1 = {
    "ALA": "A", "ARG": "R", "ASN": "N", "ASP": "D", "CYS": "C",
    "GLN": "Q", "GLU": "E", "GLY": "G", "HIS": "H", "ILE": "I",
    "LEU": "L", "LYS": "K", "MET": "M", "PHE": "F", "PRO": "P",
    "SER": "S", "THR": "T", "TRP": "W", "TYR": "Y", "VAL": "V",
    "SEC": "U", "PYL": "O",
}

def sequence_from_fasta(path):
    seq = []
    with open(path) as handle:
        for line in handle:
            line = line.strip()
            if line and not line.startswith(">"):
                seq.append(line)
    return clean_seq("".join(seq))

def parse_residue_range(residue_range):
    residue_range = str(residue_range).strip()
    if residue_range == "":
        return None, None
    if "-" not in residue_range:
        resi = int(residue_range)
        return resi, resi
    start, end = residue_range.split("-", 1)
    return int(start), int(end)

def sequence_from_pdb_chain(path, chain_id, residue_range=""):
    start_resi, end_resi = parse_residue_range(residue_range)
    residues = []
    seen = set()
    with open(path) as handle:
        for line in handle:
            if not line.startswith(("ATOM", "HETATM")) or len(line) < 27:
                continue
            chain = line[21].strip()
            if chain != chain_id:
                continue
            try:
                resi_num = int(line[22:26])
            except ValueError:
                continue
            if start_resi is not None and not (start_resi <= resi_num <= end_resi):
                continue
            resname = line[17:20].strip()
            resid = (chain, resi_num, line[26].strip())
            if resid in seen:
                continue
            seen.add(resid)
            residues.append(AA3_TO_1.get(resname, "X"))
    return clean_seq("".join(residues))

def apply_sequence_range(seq, residue_range):
    residue_range = str(residue_range).strip()
    if residue_range == "":
        return seq
    if "-" not in residue_range:
        idx = int(residue_range) - 1
        return seq[idx:idx + 1]
    start, end = residue_range.split("-", 1)
    start = int(start) - 1
    end = int(end)
    return seq[start:end]

def resolve_existing_file(path):
    path = str(path).strip()
    if os.path.isfile(path):
        return path

    basename = os.path.basename(path)
    search_dirs = []
    dirname = os.path.dirname(path)
    if dirname:
        search_dirs.append(dirname)
    if "pdb_folder" in globals():
        search_dirs.append(pdb_folder)

    seen_dirs = set()
    for folder in search_dirs:
        if not folder or folder in seen_dirs or not os.path.isdir(folder):
            continue
        seen_dirs.add(folder)
        for candidate in glob.glob(os.path.join(folder, "*")):
            if os.path.basename(candidate).lower() == basename.lower() and os.path.isfile(candidate):
                print("Resolved target sequence file:", candidate)
                return candidate

    raise FileNotFoundError(f"target_real_sequence_file not found: {path}")

def get_target_real_sequence():
    if str(target_real_sequence).strip():
        seq = clean_seq(target_real_sequence)
        print("Using target_real_sequence from text input.")
        return seq

    if str(target_real_sequence_file).strip():
        path = resolve_existing_file(target_real_sequence_file)

        lower = path.lower()
        if lower.endswith((".fa", ".fasta", ".faa", ".txt")):
            seq = apply_sequence_range(sequence_from_fasta(path), target_real_sequence_range)
        elif lower.endswith((".pdb", ".ent")):
            seq = sequence_from_pdb_chain(path, target_real_sequence_file_chain, target_real_sequence_range)
        else:
            raise ValueError("target_real_sequence_file must be FASTA, TXT, or PDB.")

        print("Using target sequence from file:", path)
        print("Target file chain:", target_real_sequence_file_chain)
        print("Target sequence range:", target_real_sequence_range if str(target_real_sequence_range).strip() else "full")
        print("Target sequence from file:", seq)
        return seq

    return ""

def split_seq_by_lengths(seq, lengths):
    seq = clean_seq(seq)
    pieces = []
    start = 0
    for length in lengths:
        pieces.append(seq[start:start + length])
        start += length
    return pieces

def format_seq_by_lengths(seq, lengths):
    return "/".join(split_seq_by_lengths(seq, lengths))

def replace_target_sequence(full_seq, chain_list, chain_lengths, target_chain, target_real_sequence):
    full_seq = clean_seq(full_seq)

    if target_real_sequence is None or str(target_real_sequence).strip() == "":
        return full_seq

    target_real_sequence = clean_seq(target_real_sequence)

    if target_chain not in chain_list:
        raise ValueError(f"target_chain {target_chain} not found in chains: {chain_list}")

    parts = split_seq_by_lengths(full_seq, chain_lengths)
    target_idx = chain_list.index(target_chain)
    expected_len = chain_lengths[target_idx]

    if len(target_real_sequence) != expected_len:
        raise ValueError(
            f"target_real_sequence length mismatch for chain {target_chain}: "
            f"got {len(target_real_sequence)}, expected {expected_len}. "
            "Use target_real_sequence_range to select the RFdiffusion target fragment."
        )

    parts[target_idx] = target_real_sequence
    return "".join(parts)

def read_heavy_atoms(pdb_path, chains):
    chain_set = set(chains)
    atoms = []
    with open(pdb_path) as handle:
        for line in handle:
            if not line.startswith(("ATOM", "HETATM")) or len(line) < 54:
                continue
            chain = line[21].strip()
            atom = line[12:16].strip()
            element = line[76:78].strip() if len(line) >= 78 else atom[0]
            if chain not in chain_set or element.upper() == "H":
                continue
            try:
                xyz = np.array([float(line[30:38]), float(line[38:46]), float(line[46:54])])
            except ValueError:
                continue
            atoms.append((chain, atom, xyz))
    return atoms

def geometry_interface_quality(pdb_path, target_chains, binder_chains, contact_cutoff=5.0, clash_cutoff=2.2):
    target_atoms = read_heavy_atoms(pdb_path, target_chains)
    binder_atoms = read_heavy_atoms(pdb_path, binder_chains)
    contacts = 0
    clashes = 0
    min_dist = math.inf

    for _, _, a in target_atoms:
        for _, _, b in binder_atoms:
            d = float(np.linalg.norm(a - b))
            min_dist = min(min_dist, d)
            if d <= contact_cutoff:
                contacts += 1
            if d <= clash_cutoff:
                clashes += 1

    return {
        "interface_contacts_5A": contacts,
        "interface_clashes_2p2A": clashes,
        "interface_min_distance": None if math.isinf(min_dist) else min_dist,
        "rosetta_available": False,
    }

def pyrosetta_interface_quality(pdb_path, target_chains, binder_chains):
    try:
        import pyrosetta
        from pyrosetta.rosetta.protocols.analysis import InterfaceAnalyzerMover
    except Exception:
        return geometry_interface_quality(pdb_path, target_chains, binder_chains)

    try:
        initialized = pyrosetta.rosetta.basic.was_init_called()
    except Exception:
        initialized = False
    if not initialized:
        pyrosetta.init("-mute all")

    pose = pyrosetta.pose_from_pdb(pdb_path)
    interface = "".join(target_chains) + "_" + "".join(binder_chains)

    try:
        mover = InterfaceAnalyzerMover(interface)
    except Exception:
        return geometry_interface_quality(pdb_path, target_chains, binder_chains)

    mover.apply(pose)

    metrics = geometry_interface_quality(pdb_path, target_chains, binder_chains)
    metrics["rosetta_available"] = True
    for key, method in [
        ("rosetta_dG_separated", "get_interface_dG"),
        ("rosetta_delta_sasa", "get_interface_delta_sasa"),
        ("rosetta_shape_complementarity", "get_interface_sc"),
        ("rosetta_packstat", "get_interface_packstat"),
    ]:
        try:
            metrics[key] = float(getattr(mover, method)())
        except Exception:
            metrics[key] = None
    return metrics

def clear_jax_memory():
    try:
        jax.clear_caches()
    except Exception:
        pass
    gc.collect()

def run_af2_ig_complex_af2_monomer_rosetta(
    max_designs_per_pdb=4,
    designed_chain="",
    complex_num_models=1,
    complex_num_recycles=1,
    monomer_num_models=1,
    monomer_num_recycles=1,
    run_monomer_af2=True,
    run_interface_quality=True,
):
    ensure_af_params(use_multimer=True)
    ensure_af_params(use_multimer=False)

    pipeline_dir = os.path.join(output_dir, "af2ig_complex_af2monomer_rosetta")
    complex_dir = os.path.join(pipeline_dir, "complex")
    monomer_dir = os.path.join(pipeline_dir, "monomer")
    tmp_dir = os.path.join(pipeline_dir, "tmp")

    for folder in [pipeline_dir, complex_dir, monomer_dir, tmp_dir]:
        os.makedirs(folder, exist_ok=True)

    result_csv = os.path.join(pipeline_dir, "af2ig_complex_af2monomer_rosetta_results.csv")
    print("Pipeline output directory:", pipeline_dir)
    print("Pipeline CSV path:", result_csv)
    target_replacement_sequence = get_target_real_sequence()
    print("Target replacement chain:", target_chain)
    print("Target replacement length:", len(target_replacement_sequence) if target_replacement_sequence else 0)

    rows = []
    completed_designs = set()
    if resume_pipeline and not force_rerun_pipeline and os.path.isfile(result_csv):
        try:
            cached_pipeline_df = pd.read_csv(result_csv)
            if len(cached_pipeline_df) > 0:
                rows = cached_pipeline_df.to_dict("records")
                completed_designs = {
                    (str(row["pdb"]), int(row["design_index"]))
                    for row in rows
                    if pd.notna(row.get("pdb")) and pd.notna(row.get("design_index")) and pd.isna(row.get("error"))
                    and (not target_replacement_sequence or (
                        str(row.get("target_real_sequence_used", "")) == target_replacement_sequence
                        and str(row.get("target_replacement_chain", "")) == target_chain
                    ))
                }
                print(f"Loaded {len(rows)} cached pipeline rows from: {result_csv}")
        except pd.errors.EmptyDataError:
            rows = []
    if force_rerun_pipeline or not os.path.isfile(result_csv):
        pd.DataFrame().to_csv(result_csv, index=False)

    for run in mpnn_runs:
        pdb_name = run["pdb_name"]
        pdb_path = run["pdb_path"]
        out = run["out"]
        selected_chains = [c for c in chains.split(",") if c]
        pdb_chains = pdb_chain_ids(pdb_path)
        chain_list = selected_chains if selected_chains else pdb_chains
        design_chain = designed_chain.strip() or chain_list[-1]
        target_chain_spec = str(globals().get("target_chains", "")).replace(" ", "")
        requested_target_chains = [c for c in target_chain_spec.split(",") if c]
        target_chains = requested_target_chains if requested_target_chains else [c for c in chain_list if c != design_chain]

        if not target_chains:
            raise ValueError("Complex scoring requires at least one target chain and one designed chain.")

        print(f"\nAF2-IG complex -> AF2 monomer -> interface quality for: {pdb_name}")
        print("Target chains:", ",".join(target_chains))
        print("Designed chain:", design_chain)

        complex_model = mk_af_model(
            use_multimer=True,
            use_templates=False,
            best_metric="i_ptm",
            initial_guess=True,
            data_dir=af_data_dir,
        )
        complex_model.prep_inputs(pdb_path, ",".join(chain_list), homooligomer=homooligomer)
        complex_lengths = list(complex_model._lengths)
        chain_to_len = dict(zip(chain_list, complex_lengths))
        design_chain_index = chain_list.index(design_chain)

        monomer_pdb = write_chain_pdb(
            pdb_path,
            design_chain,
            os.path.join(tmp_dir, f"{pdb_name}_{design_chain}.pdb")
        )

        total_designs = min(len(out["seq"]), max_designs_per_pdb)
        for n in range(total_designs):
            if (pdb_name, n) in completed_designs:
                print(f"Skipping pipeline for {pdb_name} design {n}; cached result exists.")
                continue

            original_seq = out["seq"][n]
            full_seq = replace_target_sequence(
                full_seq=original_seq,
                chain_list=chain_list,
                chain_lengths=complex_lengths,
                target_chain=target_chain,
                target_real_sequence=target_replacement_sequence,
            )
            seq_parts = split_seq_by_lengths(full_seq, complex_lengths)
            if sum(len(s) for s in seq_parts) != sum(complex_lengths):
                raise ValueError(f"Could not split designed sequence for {pdb_name} design {n}.")
            monomer_seq = seq_parts[design_chain_index]
            complex_seq = clean_seq(full_seq)[:complex_model._len]

            print(f"{pdb_name} | design {n}")
            print("Original MPNN complex sequence:", original_seq)
            print("Complex sequence after target replacement:", full_seq)
            print("Complex sequence by chain after target replacement:", format_seq_by_lengths(full_seq, complex_lengths))
            print(f"Monomer {design_chain} sequence:", monomer_seq)

            complex_pdb = os.path.join(complex_dir, f"{pdb_name}_n{n}_complex_af2ig.pdb")
            monomer_pdb_out = None
            c_log = {}
            m_log = {}
            interface_metrics = {}

            try:
                complex_model.restart()
                complex_model.predict(
                    seq=complex_seq,
                    num_recycles=complex_num_recycles,
                    num_models=complex_num_models,
                    verbose=False,
                )
                c_log = dict(complex_model.aux["log"])
                c_log["complex_confidence"] = 0.8 * c_log.get("i_ptm", 0.0) + 0.2 * c_log.get("ptm", 0.0)
                complex_model._save_results(save_best=True, best_metric="i_ptm", verbose=False)
                complex_model.save_current_pdb(complex_pdb)
                complex_model._k += 1
            except Exception as e:
                row = {
                    "pdb": pdb_name,
                    "design_index": n,
                    "target_chains": ",".join(target_chains),
                    "designed_chain": design_chain,
                    "error_stage": "af2ig_complex",
                    "error": repr(e),
                    "mpnn": out["score"][n],
                    "seqid": out["seqid"][n],
                    "seq": full_seq,
                    "complex_sequence_by_chain": format_seq_by_lengths(full_seq, complex_lengths),
                    "monomer_seq": monomer_seq,
                    "target_replacement_chain": target_chain,
                    "target_real_sequence_used": target_replacement_sequence,
                }
                rows.append(row)
                pd.DataFrame(rows).to_csv(result_csv, index=False)
                print("Saved error row to pipeline CSV:", result_csv)
                clear_jax_memory()
                continue

            clear_jax_memory()

            if run_monomer_af2:
                try:
                    monomer_model = mk_af_model(
                        use_multimer=False,
                        use_templates=False,
                        best_metric="plddt",
                        initial_guess=False,
                        data_dir=af_data_dir,
                    )
                    monomer_model.prep_inputs(monomer_pdb, design_chain, homooligomer=False)
                    monomer_model.restart()
                    monomer_model.predict(
                        seq=monomer_seq,
                        num_recycles=monomer_num_recycles,
                        num_models=monomer_num_models,
                        verbose=False,
                    )
                    m_log = dict(monomer_model.aux["log"])
                    monomer_pdb_out = os.path.join(monomer_dir, f"{pdb_name}_n{n}_{design_chain}_monomer_af2.pdb")
                    monomer_model.save_current_pdb(monomer_pdb_out)
                    del monomer_model
                    clear_jax_memory()
                except Exception as e:
                    m_log = {"error_stage": "af2_monomer", "error": repr(e)}
                    print("Monomer AF2 failed:", repr(e))
                    clear_jax_memory()

            if run_interface_quality:
                interface_metrics = pyrosetta_interface_quality(complex_pdb, target_chains, [design_chain])

            row = {
                "pdb": pdb_name,
                "design_index": n,
                "target_chains": ",".join(target_chains),
                "designed_chain": design_chain,
                "complex_pdb": complex_pdb,
                "monomer_pdb": monomer_pdb_out,
                "error_stage": m_log.get("error_stage"),
                "error": m_log.get("error"),
                "af2ig_complex_plddt": c_log.get("plddt"),
                "af2ig_complex_ptm": c_log.get("ptm"),
                "af2ig_complex_i_ptm": c_log.get("i_ptm"),
                "af2ig_complex_confidence": c_log.get("complex_confidence"),
                "af2_monomer_plddt": m_log.get("plddt"),
                "af2_monomer_ptm": m_log.get("ptm"),
                "mpnn": out["score"][n],
                "seqid": out["seqid"][n],
                "seq": full_seq,
                "complex_sequence_by_chain": format_seq_by_lengths(full_seq, complex_lengths),
                "monomer_seq": monomer_seq,
                "target_replacement_chain": target_chain,
                "target_real_sequence_used": target_replacement_sequence,
            }
            row.update(interface_metrics)
            rows.append(row)

            # Save after every design so partial results survive interruptions.
            pd.DataFrame(rows).to_csv(result_csv, index=False)
            print("Updated pipeline CSV:", result_csv)
            clear_jax_memory()

    df = pd.DataFrame(rows)
    df.to_csv(result_csv, index=False)
    print("Saved pipeline results:", result_csv)
    return df

if run_pipeline_now:
    pipeline_df = run_af2_ig_complex_af2_monomer_rosetta(
        max_designs_per_pdb=max_designs_per_pdb,
        designed_chain=designed_chain,
        complex_num_models=complex_num_models,
        complex_num_recycles=complex_num_recycles,
        monomer_num_models=monomer_num_models,
        monomer_num_recycles=monomer_num_recycles,
        run_monomer_af2=run_monomer_af2,
        run_interface_quality=run_interface_quality,
    )
    data_table.DataTable(
        pipeline_df.sort_values(
            ["af2ig_complex_confidence", "af2_monomer_plddt"],
            ascending=[False, False]
        ).round(3)
    )
```

```python
#@title Install / Verify PyRosetta

#@markdown Run this before the FastRelax cell.
install_pyrosetta_now = False #@param {type:"boolean"}
pyrosetta_build = "release" #@param ["release", "minsizerel"]
pyrosetta_mirror = "west" #@param ["west", "east"]

import os, sys, subprocess, importlib

def install_or_verify_pyrosetta(
    install_now=False,
    build="release",
    mirror="west",
):
    try:
        import pyrosetta
        pyrosetta.init("-mute all")
        print("PyRosetta already installed.")
        print("PyRosetta module:", pyrosetta.__file__)
        return pyrosetta
    except Exception as first_error:
        print("PyRosetta import/init failed:", repr(first_error))

    if not install_now:
        print("Set install_pyrosetta_now=True and rerun this cell to install PyRosetta.")
        return None

    mirror_host = {
        "west": "https://west.rosettacommons.org",
        "east": "https://east.rosettacommons.org",
    }.get(mirror, "https://west.rosettacommons.org")

    build_path = {
        "release": "quarterly/release",
        "minsizerel": "quarterly/minsizerel",
    }.get(build, "quarterly/release")

    find_links = f"{mirror_host}/pyrosetta/{build_path}"

    print("Installing PyRosetta from:", find_links)
    print("This download can be large and may take several minutes.")

    try:
        subprocess.check_call([
            sys.executable,
            "-m",
            "pip",
            "install",
            "-q",
            "pyrosetta",
            "--find-links",
            find_links,
        ])
    except Exception as pip_error:
        print("Direct mirror install failed:", repr(pip_error))
        print("Trying pyrosetta-installer fallback.")
        subprocess.check_call([
            sys.executable,
            "-m",
            "pip",
            "install",
            "-q",
            "pyrosetta-installer",
        ])
        import pyrosetta_installer
        pyrosetta_installer.install_pyrosetta()

    importlib.invalidate_caches()
    import pyrosetta
    pyrosetta.init("-mute all")

    print("PyRosetta installed and initialized.")
    print("PyRosetta module:", pyrosetta.__file__)
    return pyrosetta

pyrosetta = install_or_verify_pyrosetta(
    install_now=install_pyrosetta_now,
    build=pyrosetta_build,
    mirror=pyrosetta_mirror,
)
```

```python
#@title Rosetta FastRelax on AF2 Complex Outputs

#@markdown Run this after the AF2-IG pipeline cell.
run_fastrelax_now = False #@param {type:"boolean"}
max_relax_models = 10 #@param {type:"integer"}
relax_repeats = 1 #@param {type:"integer"}
resume_fastrelax = True #@param {type:"boolean"}
force_rerun_fastrelax = False #@param {type:"boolean"}

import os, math
import numpy as np
import pandas as pd

if "output_dir" not in dir():
    output_dir = os.path.join(
        "/content/drive/MyDrive/sequence_input_diffusion_example_outputs",
        "proteinmpnn_alphafold_results"
    )

pipeline_dir = os.path.join(output_dir, "af2ig_complex_af2monomer_rosetta")
pipeline_csv = os.path.join(pipeline_dir, "af2ig_complex_af2monomer_rosetta_results.csv")
relax_dir = os.path.join(pipeline_dir, "fastrelax")
relax_csv = os.path.join(pipeline_dir, "fastrelax_interface_results.csv")

os.makedirs(relax_dir, exist_ok=True)

def read_heavy_atoms_for_relax(pdb_path, chains):
    chain_set = set(chains)
    atoms = []
    with open(pdb_path) as handle:
        for line in handle:
            if not line.startswith(("ATOM", "HETATM")) or len(line) < 54:
                continue
            chain = line[21].strip()
            atom = line[12:16].strip()
            element = line[76:78].strip() if len(line) >= 78 else atom[0]
            if chain not in chain_set or element.upper() == "H":
                continue
            try:
                xyz = np.array([float(line[30:38]), float(line[38:46]), float(line[46:54])])
            except ValueError:
                continue
            atoms.append((chain, atom, xyz))
    return atoms

def fallback_interface_quality_relax(pdb_path, target_chains, binder_chains, contact_cutoff=5.0, clash_cutoff=2.2):
    target_atoms = read_heavy_atoms_for_relax(pdb_path, target_chains)
    binder_atoms = read_heavy_atoms_for_relax(pdb_path, binder_chains)
    contacts = 0
    clashes = 0
    min_dist = math.inf

    for _, _, a in target_atoms:
        for _, _, b in binder_atoms:
            d = float(np.linalg.norm(a - b))
            min_dist = min(min_dist, d)
            if d <= contact_cutoff:
                contacts += 1
            if d <= clash_cutoff:
                clashes += 1

    return {
        "relaxed_interface_contacts_5A": contacts,
        "relaxed_interface_clashes_2p2A": clashes,
        "relaxed_interface_min_distance": None if math.isinf(min_dist) else min_dist,
    }

def run_fastrelax_one_pdb(input_pdb, output_pdb, target_chains, binder_chains, relax_repeats=1):
    try:
        import pyrosetta
        from pyrosetta.rosetta.protocols.relax import FastRelax
        from pyrosetta.rosetta.protocols.analysis import InterfaceAnalyzerMover
    except Exception as e:
        raise ImportError(
            "PyRosetta is required for FastRelax. Install/import PyRosetta first. "
            "The fallback geometry metrics do not perform relaxation."
        ) from e

    try:
        initialized = pyrosetta.rosetta.basic.was_init_called()
    except Exception:
        initialized = False

    if not initialized:
        pyrosetta.init("-mute all")

    pose = pyrosetta.pose_from_pdb(input_pdb)
    scorefxn = pyrosetta.get_fa_scorefxn()

    relax = FastRelax()
    relax.set_scorefxn(scorefxn)
    relax.max_iter(200)

    for _ in range(max(1, int(relax_repeats))):
        relax.apply(pose)

    pose.dump_pdb(output_pdb)

    interface = "".join(target_chains) + "_" + "".join(binder_chains)
    metrics = {
        "relaxed_pdb": output_pdb,
        "relaxed_total_score": float(scorefxn(pose)),
        "fastrelax_available": True,
    }

    try:
        mover = InterfaceAnalyzerMover(interface)
        mover.apply(pose)
        for key, method in [
            ("relaxed_rosetta_dG_separated", "get_interface_dG"),
            ("relaxed_rosetta_delta_sasa", "get_interface_delta_sasa"),
            ("relaxed_rosetta_shape_complementarity", "get_interface_sc"),
            ("relaxed_rosetta_packstat", "get_interface_packstat"),
        ]:
            try:
                metrics[key] = float(getattr(mover, method)())
            except Exception:
                metrics[key] = None
    except Exception as e:
        metrics["fastrelax_interface_error"] = repr(e)

    metrics.update(fallback_interface_quality_relax(output_pdb, target_chains, binder_chains))
    return metrics

def run_fastrelax_pipeline(max_relax_models=10, relax_repeats=1):
    if not os.path.isfile(pipeline_csv):
        raise FileNotFoundError(f"Pipeline CSV not found: {pipeline_csv}")

    source_df = pd.read_csv(pipeline_csv)
    if len(source_df) == 0:
        raise ValueError(f"Pipeline CSV is empty: {pipeline_csv}")

    rows = []
    completed = set()

    print("FastRelax input CSV:", pipeline_csv)
    print("FastRelax output directory:", relax_dir)
    print("FastRelax result CSV:", relax_csv)

    if resume_fastrelax and not force_rerun_fastrelax and os.path.isfile(relax_csv):
        try:
            cached = pd.read_csv(relax_csv)
            rows = cached.to_dict("records")
            completed = {
                (str(row["pdb"]), int(row["design_index"]))
                for row in rows
                if pd.notna(row.get("pdb")) and pd.notna(row.get("design_index")) and pd.isna(row.get("fastrelax_error"))
            }
            print(f"Loaded {len(rows)} cached FastRelax rows.")
        except pd.errors.EmptyDataError:
            rows = []

    candidates = source_df.copy()
    if "error" in candidates.columns:
        candidates = candidates[candidates["error"].isna()]
    if "complex_pdb" in candidates.columns:
        candidates = candidates[candidates["complex_pdb"].notna()]

    if "af2ig_complex_confidence" in candidates.columns:
        candidates = candidates.sort_values("af2ig_complex_confidence", ascending=False)

    candidates = candidates.head(max_relax_models)

    for _, row in candidates.iterrows():
        pdb_name = str(row["pdb"])
        design_index = int(row["design_index"])

        if (pdb_name, design_index) in completed:
            print(f"Skipping FastRelax for {pdb_name} design {design_index}; cached result exists.")
            continue

        input_pdb = str(row["complex_pdb"])
        if not os.path.isfile(input_pdb):
            result = row.to_dict()
            result.update({
                "fastrelax_available": False,
                "fastrelax_error": f"Missing complex PDB: {input_pdb}",
            })
            rows.append(result)
            pd.DataFrame(rows).to_csv(relax_csv, index=False)
            continue

        target = str(row["target_chains"]).split(",")
        binder = [str(row["designed_chain"])]
        output_pdb = os.path.join(relax_dir, f"{pdb_name}_n{design_index}_fastrelax.pdb")

        print(f"FastRelax: {pdb_name} design {design_index}")
        print("Input:", input_pdb)
        print("Output:", output_pdb)

        result = row.to_dict()
        try:
            result.update(
                run_fastrelax_one_pdb(
                    input_pdb=input_pdb,
                    output_pdb=output_pdb,
                    target_chains=target,
                    binder_chains=binder,
                    relax_repeats=relax_repeats,
                )
            )
            result["fastrelax_error"] = None
        except Exception as e:
            result.update(fallback_interface_quality_relax(input_pdb, target, binder))
            result["fastrelax_available"] = False
            result["fastrelax_error"] = repr(e)
            result["relaxed_pdb"] = None
            print("FastRelax failed:", repr(e))

        rows.append(result)
        pd.DataFrame(rows).to_csv(relax_csv, index=False)
        print("Updated FastRelax CSV:", relax_csv)

    df = pd.DataFrame(rows)
    df.to_csv(relax_csv, index=False)
    print("Saved FastRelax results:", relax_csv)
    return df

if run_fastrelax_now:
    fastrelax_df = run_fastrelax_pipeline(
        max_relax_models=max_relax_models,
        relax_repeats=relax_repeats,
    )
    data_table.DataTable(
        fastrelax_df.sort_values(
            ["relaxed_rosetta_dG_separated", "relaxed_interface_clashes_2p2A"],
            ascending=[True, True],
            na_position="last",
        ).round(3)
    )
```
